### gradient descent for linear regression

```python
def gradient_descent(x, y, lr=0.01, epochs= 100, batch_size=10):
    n = len(y)
    w = np.random.randn()
    b = np.random.randn()
    
    for epoch in range(epochs):
        indices = np.arange(n)
        np.random.shuffle(indices)
        
        for start in range(0, n, batch_size):
            end = start + batch_size
            batch_indices = indices[start:end]
            x_batch = x[batch_indices]
            y_batch = y[batch_indices]
            
            y_pred = w * x_batch + b
            error = y_pred - y_batch
            
            dw = (2/batch_size) * np.dot(error, x_batch)
            db = (2/batch_size) * np.sum(error)
            
            w -= lr * dw
            b -= lr * db
            
    return w, b
```

```python
def full_batch_gradient_descent(x, y, lr=0.01, epochs= 100):
    w, b = gradient_descent(x, y, lr=lr, epochs=epochs, batch_size=len(y))
    return w, b
```

```python
def stochastic_gradient_descent(x, y, lr=0.01, epochs= 100):
    w, b = gradient_descent(x, y, lr=lr, epochs=epochs, batch_size=1)
    return w, b
```

True value of w and b using normal equation

```python
# add intercept, X = [x, 1]
X = np.column_stack((x1, np.ones_like(x1)))

#  theta = (X^T X)^(-1) X^T y
theta = np.linalg.inv(X.T @ X) @ X.T @ y

w, b = theta

print("w =", w)
print("b =", b)
```
